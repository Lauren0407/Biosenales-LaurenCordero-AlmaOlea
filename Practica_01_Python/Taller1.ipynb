### Punto P

```python
def escalón(n_0, n_1, n_2):    #Función escalón unitario definida para el intervalo  𝑛1 ≤ 𝑛0 ≤ 𝑛2
    n = np.arange(n_1, n_2+1)
    x= np.where(n>= n_0, 1, 0)
    return n,x

#Ejemplo de uso
n0, n1, n2 = 1, -5, 5
n, x = escalón(n0, n1, n2)

# Gráfica
plt.plot(n, x, color='purple', linestyle='-', marker='*')
plt.title(f'Escalón unitario u(n - {n0})')
plt.xlabel('n')
plt.ylabel('u(n - n0)')
plt.grid(True)
plt.show()
```

### Punto Q

```python
def rampa(n_0, n_1, n_2, m): #Función rampa definida en un intervalo y una pendiente m
    n = np.arange(n_1, n_2 + 1)
    x = np.where(n >= n_0, m * (n - n_0), 0)
    return n, x

# Ejemplo de uso
n0, n1, n2 = 0, -5, 10
m = 4  # Valor dado para la pendiente

n, x = rampa(n0, n1, n2, m)

# Gráfica
plt.plot(n, x, color='green', linestyle='-', marker='+')
plt.title(f'Rampa r(n - {n0}) con pendiente m={m}')
plt.xlabel('n')
plt.ylabel('r(n - n0)')
plt.grid(True)
plt.show()
```

### Punto S

#### Punto S-a

```python
def rampa(n_0, n_1, n_2, m=1):
    n = np.arange(n_1, n_2 + 1)
    x = np.where(n >= n_0, m * (n - n_0), 0)
    return n, x

def r(valor):
    return valor if valor >= 0 else 0

# Se genera x[n] para n = 0 hasta 17
n = np.arange(0, 18)
x = np.zeros(len(n))

for i, ni in enumerate(n):
    if 0 <= ni <= 5:
        x[i] = r(ni)
    elif 6 <= ni <= 11:
        x[i] = r(ni - 5)
    elif 12 <= ni <= 17:
        x[i] = r(ni - 10)

#Se deriva la secuencia 
dx = np.diff(x, prepend=x[0])  # dx[n] = x[n] - x[n-1]
```

#### Punto S-b

```python
plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.stem(n, x)
plt.title('Secuencia original x[n]')
plt.xlabel('n')
plt.ylabel('x[n]')
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(n, dx)
plt.title('Derivada discreta dx[n] = x[n] - x[n-1]')
plt.xlabel('n')
plt.ylabel('dx[n]')
plt.grid(True)

plt.tight_layout()
plt.show()
```

### Punto 1

```python
def notas_estudiantes(diccionario):  #Función para recibiri un diccionario con las notas de los alumnos de un curso y entregar valores estadísticos
    notas = list(diccionario.values())
    
    resultado = pd.Series({
        'Nota mínima': np.min(notas),
        'Nota máxima': np.max(notas),
        'Media estadística': np.mean(notas),
        'Desviación típica': np.std(notas)})
    return resultado

# Ejemplo de uso
notas_curso = {
    'Alma': 4.5,
    'Lauren': 3.8,
    'Rosa': 5.0,
    'María': 2.87,
    'Juliana': 1.0}

resultado_obtenido = notas_estudiantes(notas_curso)
print(resultado_obtenido)
```
